```python
# Підготовка бібліотеки NumPy.
import numpy as np
np.random.seed(25)
```

```python
# Завдання 1. Створіть одновимірний масив (вектор) з сумами добутків двох попередніх чисел та їхнього суми, починаючи з 1, та виведіть його значення.
# Умова не задає довжину та друге початкове число. Беру 10 елементів, a[0] = a[1] = 1.
a = [1, 1]
for i in range(2, 10):
    a.append(a[-2] * a[-1] + a[-2] + a[-1])
a = np.array(a, dtype=object)
print(a)
```

```python
# Завдання 2. Створіть двовимірний масив (матрицю) розміром 3 x3, заповніть його значеннями випадкових цілих чисел від -10 до 10.
a = np.random.randint(-10, 11, (3, 3))
print(a)
```

```python
# Завдання 3. Створіть масив розміром 5x5, заповніть його випадковими дійсними числами від 0 до 1 та знайдіть відсоток чисел, які є простими.
a = np.random.uniform(0, 1, (5, 5))
prime_count = 0
for x in a.flat:
    if x >= 2 and float(x).is_integer():
        n = int(x)
        if all(n % d != 0 for d in range(2, int(n ** 0.5) + 1)):
            prime_count += 1
print(a)
print('Відсоток простих чисел:', prime_count / a.size * 100)
# Прості числа є цілими та більшими за 1, тому в діапазоні [0, 1) їх немає.
```

```python
# Завдання 5. Створіть два одновимірних масиви розміром 4, заповніть їх випадковими цілими числами в діапазоні від 15 до 25 та виконайте на них поелементні операції додавання, віднімання та множення.
a = np.random.randint(15, 26, 4)
b = np.random.randint(15, 26, 4)
print('a:', a, '\nb:', b)
print('Додавання:', a + b)
print('Віднімання:', a - b)
print('Множення:', a * b)
```

```python
# Завдання 6. Створіть дві матриці 3 x3, заповніть їх довільними цілими числами від 1 до 10 та знайдіть їх добуток.
a = np.random.randint(1, 11, (3, 3))
b = np.random.randint(1, 11, (3, 3))
print('a:', a, '\nb:', b)
print('Добуток:', a @ b)
```

```python
# Завдання 7. Створіть дві матриці розміром 2x3 та 3x3, заповніть їх випадковими цілими числами в діапазоні від 1 до 5 та перемножте їх між собою.
a = np.random.randint(1, 6, (2, 3))
b = np.random.randint(1, 6, (3, 3))
print('a:', a, '\nb:', b)
print('Добуток:', a @ b)
```

```python
# Завдання 8. Створіть матрицю розміром 2x2, заповніть її випадковими цілими числами в діапазоні від 1 до 7 та знайдіть її обернену матрицю.
a = np.random.randint(1, 8, (2, 2))
while np.linalg.matrix_rank(a) < 2:
    a = np.random.randint(1, 8, (2, 2))
print(a)
print('Обернена матриця:', np.linalg.inv(a))
```

```python
# Завдання 9. Створіть матрицю розміром 3 x3, заповніть її випадковими дійсними числами в діапазоні від 5 до 10 та транспонуйте її.
a = np.random.uniform(5, 10, (3, 3))
print(a)
print('Транспонована матриця:', a.T)
```

```python
# Завдання 12. Створіть дві матриці розміром 2 x2, заповніть їх випадковими цілими числами в діапазоні від -2 до 2 та знайдіть суму всіх елементів обох матриць.
a = np.random.randint(-2, 3, (2, 2))
b = np.random.randint(-2, 3, (2, 2))
print('a:', a, '\nb:', b)
print('Сума всіх елементів:', a.sum() + b.sum())
```

```python
# Завдання 13. Створіть дві матриці розміром 3x3, заповніть їх випадковими дійсними числами в діапазоні від 0 до 1 та знайдіть суму їх елементів.
a = np.random.uniform(0, 1, (3, 3))
b = np.random.uniform(0, 1, (3, 3))
print('a:', a, '\nb:', b)
print('Поелементна сума:', a + b)
print('Сума елементів a:', a.sum())
print('Сума елементів b:', b.sum())
print('Загальна сума:', a.sum() + b.sum())
```

```python
# Завдання 14. Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні від -10 до 10 та знайдіть суму елементів на побічній діагоналі.
a = np.random.uniform(-10, 10, (3, 3))
print(a)
print('Сума побічної діагоналі:', np.fliplr(a).trace())
```

```python
# Завдання 15. Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні від 0 до 1 та знайдіть вектор -стовпчик, що містить найбільший елемент у кожному стовпці матриці.
a = np.random.uniform(0, 1, (3, 3))
b = a.max(axis=0).reshape(-1, 1)
print(a)
print('Вектор-стовпчик максимумів:', b)
```
